```python
import requests

data = requests.get(
    "https://codeforces.com/api/problemset.problems"
).json()

problems = data["result"]["problems"]

questions = []

for p in problems:
    rating = p.get("rating")

    if p["tags"] == ["dp"] and rating and 800 <= rating <= 2000:
        link = f"https://codeforces.com/problemset/problem/{p['contestId']}/{p['index']}"

        questions.append((
            rating,
            p["name"],
            link
        ))

questions.sort()

for rating, name, link in questions:
    print(f"{rating} | {name}")
    print(link)
```

```python
import requests

data = requests.get(
    "https://codeforces.com/api/problemset.problems"
).json()

problems = data["result"]["problems"]

questions = []

for p in problems:
    rating = p.get("rating")

    if set(p["tags"]) == {"dp", "bitmasks"} and rating and 800 <= rating <= 2000:
        link = f"https://codeforces.com/problemset/problem/{p['contestId']}/{p['index']}"

        questions.append((
            rating,
            p["name"],
            link
        ))

questions.sort()

for rating, name, link in questions:
    print(f"{rating} | {name}")
    print(link)
```
